# Outcome Dynamics — individual trajectories

Single simulation without treatment. Two plots: one per outcome, all firm trajectories + mean.

```python
import sys
import copy
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, '..')

from data_generation.simulator import DataSimulator
from data_generation.data_config import DATA_CONFIG

import pandas as pd

import warnings
warnings.filterwarnings("ignore", category=pd.errors.PerformanceWarning)
```

```python
# Parámetros fijos
N_FIRMS   = 50
N_PERIODS = 1000
SEED = 2024
```

```python
cfg = copy.deepcopy(DATA_CONFIG)
cfg.update({
    'n_empresas':                N_FIRMS,
    'n_periodos':                N_PERIODS,
    'n_cohortes':                0,
    'cupo_por_cohorte':          [],
    'periodo_inicio_programa':   N_PERIODS,
    'min_periodos_pre_programa': N_PERIODS,
    'random_seed':               SEED,
})

sim   = DataSimulator(cfg)
panel = sim.simulate()
print(f"{panel['id_firma'].nunique()} firms, {panel['t'].nunique()} periods")
```

```python
outcomes = list(DATA_CONFIG['dinamica_outcomes'].keys())

for outcome in outcomes:
    fig, ax = plt.subplots(figsize=(10, 5))

    for _, firm_df in panel.groupby('id_firma'):
        ax.plot(firm_df['t'], firm_df[outcome], color='steelblue', alpha=0.15, linewidth=0.8)

    mean_traj = panel.groupby('t')[outcome].mean()
    ax.plot(mean_traj.index, mean_traj.values, color='crimson', linewidth=2.2, label='mean')

    dyn = DATA_CONFIG['dinamica_outcomes'][outcome]
    ax.set_title(
        f"{outcome}\npersistencia={dyn['persistencia']}, volatilidad={dyn['volatilidad']}",
        fontsize=12
    )
    ax.set_xlabel('Período (t)', fontsize=11)
    ax.set_ylabel(outcome, fontsize=11)
    ax.legend(fontsize=10)
    ax.grid(True, alpha=0.3)

    fig.suptitle(
        f'Dinámica de outcomes — sin tratamiento\nn={N_FIRMS}, T={N_PERIODS}, seed={SEED}',
        fontsize=13
    )
    plt.tight_layout()
    plt.show()
```
